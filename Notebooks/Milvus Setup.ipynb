```python
from pymilvus import CollectionSchema, FieldSchema, DataType, Collection, connections
```

```python

# Connect to the Milvus server
connections.connect("default", host="localhost", port="19530")

# Check connection
if connections.has_connection("default"):
    print("Connected to Milvus successfully!")
else:
    print("Failed to connect to Milvus.")
```

```python

# Define the schema for the collection
sentence_id = FieldSchema(name="sentence_id", dtype=DataType.INT64, is_primary=True, auto_id=True)
document_id = FieldSchema(name="document_id", dtype=DataType.VARCHAR, max_length=512)
sentence_embedding = FieldSchema(name="sentence_embedding", dtype=DataType.FLOAT_VECTOR, dim=768)
sentence_text = FieldSchema(name="sentence_text", dtype=DataType.VARCHAR, max_length=1024)

# Create the collection schema
schema = CollectionSchema(fields=[sentence_id, document_id, sentence_embedding, sentence_text], description="Collection of sentence embeddings")

# Create the collection in Milvus
collection = Collection(name="sentence_embeddings", schema=schema)
```
