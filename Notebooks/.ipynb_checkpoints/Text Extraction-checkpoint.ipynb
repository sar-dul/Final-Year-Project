```python
import os
import zipfile
from pathlib import Path
from PyPDF2 import PdfReader
```

```python
def process_file(file_path):
    # Get the extension of the submitted file
    file_extension = Path(file_path).suffix

    if file_extension == '.zip':
        # Extract files from the zip archive
        with zipfile.ZipFile(file_path, 'r') as zip_ref:
            extract_dir = "extracted_files"
            zip_ref.extractall(extract_dir)
            print(f"Extracted all files to {extract_dir}")

        # Get all PDF and Python (.py) files
        file_contents = {}
        for root, dirs, files in os.walk(extract_dir):
            for file in files:
                if file.endswith('.pdf') or file.endswith('.py'):
                    file_full_path = os.path.join(root, file)
                    content = store_file(file_full_path)
                    file_contents[file] = content
        return file_contents

    elif file_extension == '.pdf' or file_extension == '.py':
        # If it's a .pdf or .py file, store its content directly
        content = store_file(file_path)
        return {file_path: content}
    else:
        print("Unsupported file type.")
        return None

def store_file(file_path):
    # Check file extension to determine the handling method
    if file_path.endswith('.pdf'):
        return extract_pdf_text(file_path)
    elif file_path.endswith('.py'):
        return read_python_file(file_path)

def extract_pdf_text(pdf_path):
    # Read the content from the PDF file
    reader = PdfReader(pdf_path)
    text = ''
    for page in reader.pages:
        text += page.extract_text()
    print(f"Extracted text from PDF: {pdf_path}")
    return text

def read_python_file(file_path):
    # Read the Python (.py) file content
    with open(file_path, 'r', encoding='utf-8') as file:
        content = file.read()
    print(f"Stored content of Python file: {file_path}")
    return content
```

```python
# Example usage
file_path = 'test.py'  # Replace with actual file path
file_contents = process_file(file_path)
print(file_contents)  # This will contain all stored content in variables
```
